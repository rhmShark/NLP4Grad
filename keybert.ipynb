```python
import re

reviews = [
    "Great stay! Booked two rooms for four adults and one kid.",
    "Traveled as 2 adults in 1 deluxe room, very clean.",
    "Amazing service for our group of 3 adults."
]

def extract_booking_details(text):
    adults = re.search(r'(\d+)\s*(?:adult|adults|guest|guests)', text, re.IGNORECASE)
    rooms = re.search(r'(\d+)\s*(?:room|rooms|suite|suites)', text, re.IGNORECASE)
    children = re.search(r'(\d+)\s*(?:child|children|kid|kids)', text, re.IGNORECASE)
    
    return {
        "adults": int(adults.group(1)) if adults else None,
        "rooms": int(rooms.group(1)) if rooms else None,
        "children": int(children.group(1)) if children else None,
    }

for r in reviews:
    print(extract_booking_details(r))
```
